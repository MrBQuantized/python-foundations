# Vectorization (Loop-less/Efficient coding)
---

### What is Vectorization?
Before we talk about this concept, we need to discuss programming languages with respect to characteristics - Intepreted or compiled

1. **Interpreted**  — each statement is anlayzed once it is executed. Takes a lot of time, especially if there are lots of statements in the program.
Examples; Python, R, PHP,..

2. **Compiled** — faster because all the code is compiled and all analysis is done prehand, just before running the code. E.g C

Python will take up more time to execute if there are multiple statements in the program compared to C. In the loop as shown below, there are lot of statements that get executed when unravelled even though it may appear to be one statement-multiple statements are repeated.
```python
for i in range(10000):
    print(i)
```
For vectorization, the whole point is to reduce the number of statements that are interpreted. So, we can define vectorization as a combination of the following:
- Avoiding explicit loops whenever possible.
- Transforming loops to array expressions. Python has built-in functions in modules such as **Numpy**  that perform operations faster than a hard-coded loop. Numpy has functions which has loops (written in a compiler making them very much faster).




### Implementation

````python
import numpy as np
````

````python
X = np.random.rand(1, 100000)
Y = np.random.rand(1, len(X))
````

````python
# Testing with Dot product
# Function with a loop
def dot_product(X,Y):
    s = 0
    for i in range(len(X)):
        s += X[i]*Y[i]
    return s

# Vectorized function
def vecDot(X,Y):
    return (X*Y).sum()
````

````python
#Let's time our functions
%timeit dot_product(X,Y)
````

````python
%timeit vecDot(X,Y)
````

We can see the difference in the time taken for executing the unvecotrized and the vectorized codes;

The vectorized is 8 times faster than the unvectorized one on average.

#### IPython Magic command

`%timeit` is a special/magic IPython command that powers Jupyter Notebook and Jupyter Lab. Another commmon command is `%matplotlib inline`.

- `%timeit` → measures how long a piecee of Python code takes to run.
- `%matplotlib inline` → tells Jupyter to display Matplot plots directly inside the notebook.

There are two main types of these magic commands:
1. **Line magics** — `%`. They operate on one lines
2. **Cell magics** — `%%`. They operate on an entire cell:

````python
# Frobannious
# random.rannd creates an array of the given shape and populate it 
# with random samples from a uniform distribution over [
X = np.random.rand(10, 30, 100, 20)  # A tensor
Y = np.random.rand(10, 30, 100, 20)  

X.shape, Y.shape
````

````python
# Square differences 

# Looped function

def square_diff(X,Y):  
    s = 0

    # 4 nested for-loops for a 4D matrix (Tensor)
    for i in range(X.shape[0]):
        for j in range(X.shape[1]):
            for k in range(X.shape[2]):
                for l in range(X.shape[3]):
                    s += (X[i,j,k,l]-Y[i,j,k,l])**2

    return s


# Vecotrized function
def vecSumDiff(X,Y):
    s = np.sum((X.flatten()-Y.flatten())**2)    # works either ways
    s = np.sum((X-Y)**2)

    return s
````

````python
%timeit square_diff(X,Y)
````

````python
%timeit vecSumDiff(X,Y)
````
