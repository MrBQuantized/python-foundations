# Pathlib Module
---
This module is used in place of the `os module`for working with paths. It was introduced in the Standard library in python 3.4 and provides and easy way to work with file paths and file system.

## How it differs from the os module

- In the os module you're working with strings for path which can get messy especially when working between different Operating Systems.

- Pathlib represents file paths as objects rather strings. It will better handle those problems a bit more
- It is more intuitive, readable and less prone to errors.


Let's take a look at the two lines from a Django project below:
- The first is from an older django project and uses the os module and builds paths like this:
  ```python
  os.path.join(BASE_DIR, 'subdir')
  ```
- The second is from a newer django project and makes use of the pathlib module. It builds paths like this:
  ```python
  BASE_DIR / 'subdir'.


````python
import os
from pathlib import Path
````

````python

BASE_DIR = os.path.dirname(os.path.dirname(os.path.abspath(__file__)))
````

From the example above, they're basically trying to set the base directory to the main project directory which is two levels up

>"***directory of the directory of the absolute path of `__file__`.***

You can see how complicated things are and it is a bit confusing to know what's going on except you break it down slowly.

Now compared to Pathlib below, the syntax is a bit simpler. We're resolving the absolute path of the variable `"__file__"` and access the parent(parent) of the path.

````python
BASE_DIR = Path(__file__).resolve().parent.parent
````

Now let's treat more examples of when to use pathlib, os module or a better module.

````python
print(Path.cwd())   # defaults to the cwd if no arguments is given
````

Let's print the files and folders within the cwd by using the iteration method

````python
for p in Path().iterdir():
    print(p)
````

## Relative Path

> Starts from the current working directory `cwd`.

````python
my_dir = Path("pathlib_os")
my_file = Path("file_1.txt")

print(my_dir)
print(my_file)
````

To be more explicit you can print out the variables with their `'name'` attributes

````python
print(my_dir.name)
print(my_file.name)
````

We can also print out th file extensions by using the `suffix` attribute provided by the path module.
This is much easier compared to the os module where you have to split the file name into the name and extension and then grab the index of the extension (or any of the items you want to display).

````python
print(my_dir.suffix)
print(my_file.suffix)
````

As expected, there's no extension for the directory but, for our text file, we have the `'.txt'` displayed.

In addition, getting the file name without the extension is called the `'stem'`.

````python
print(my_file.stem)
````

### Accessing file within a Directory
Using the Path module we can access a file in any directory and the file doesn't need to to exist first before we do so. we can create a new file and even much easier than using the os module.

A forward slash operator `/` is used in creating the file

````python
new_file = my_dir / "new_file"          # We make use of the directory variable created

print(new_file)
````

We can do the same using the  os module by using the `.joinpath()` method

````python
new_file = my_dir.joinpath("new_file.txt")

print(new_file)
````

Recall that the file `new_file.txt` doesn't need to exist before we can treat it as such. In oder to check the existence of a file, we can use the `.exists()` method  

````python
print(my_dir.exists())
print(my_file.exists()) 
print(new_file.exists())
````

### Absolute vs Relative path
Let't begin by checking the parent directory of the following files using the `.parent` attribute

````python
print(my_dir.parent)
print(my_file.parent) 
print(new_file.parent)
````

The dot `(.)` from the first two paths indicates that the cwd is the parent directory of those paths and `pathlib_os` is the parent dir of `new_file`.  We can do the same for `my_file`by adding another parent attribute (going up 2 levels) so that it becomes. `.parent.parent`

````python
print(my_dir.parent)
print(my_file.parent) 
print(new_file.parent.parent)
````

## Absolute Path
> Starts from the root of your drive/computer. It works no matter where your script is running.

In order for us to know where exactly we are and have an absolute path to these locations, there are two methods for this:
1. Using the `.absolute()`  method and the parent object. The absolute method can be used either on the parent object or the original object
   ```python
   my_dir.parent.absolute()
   # or
   my_dir.absolute().parent
   ```
    It produces the same result.

````python
print(my_dir.parent.absolute())
print(my_dir.absolute().parent) 
````

we can also get absolute path without the parent file by removing the parent object

````python
print(my_dir.absolute())
````

The second way of getting the absolute path is to use the `resolve()` method. This also resolves any relative directory references

````python
print(my_dir.resolve())
````

We do not see any difference because there are no relative references here.  However, we can create another file just to for illustration purposes.

````python
p = Path("..").absolute()
p
````

`(..)` is a common relative directory reference which usually means the parent directory. But using the `absolute()` method doesn't resolve this. 

Now let's check with the `resolve()` method

````python
p = Path("..").resolve()   
p
````

You can see that it resolves the relative dir reference by removing the (..).

Now let's do the same with Django project introduced earlier.

````python
p = Path(__file__).resolve()
p 
````

We are getting this error because the file `__file__` doesn't exist in 3 places:
1. Python Interactive  / REPL - the >>> prompt
2. Jupyter Notebook / Google Colab. 
3. IDLE's shell window.

It only works in a .py script file. It is also worthwhile to note that Jupyter runs cells, not a `.py` script.

However, we can fix this on Jupyter notebooks by using `Path.cwd() + .resolve()` to force it to absolute and remove relative bits.

````python
p = Path.cwd().resolve()
p
````

Recall that `.resolve()` does 2 things:
1. Converts it to absolute C:\Users\Dell\Documents\...
2. Removes any `...` or`.` if they exist

Other useful ones are 
1. Absolute path to a file/subfolder

````python
p = Path.cwd().resolve() / "new_file.txt"
p
````

2. Absolute path to a parent folder

````python
p = Path.cwd().resolve().parent
p
````

3. Absolute path from any string

````python
p = Path("./script.py").resolve()
p
````

Another caveat about getting an absolute path that sometimes, can't be taken care off by `resolve()`. 

For instance, I move to my home folder where my `dotfiles` are, and the symbol, tilde `~` represents my `Users` home folder,

````python
p = Path("~/dotfiles").resolve()
p
````

this doesn't give the result that we want, because it only appends `dotfiles` to the cwd. We can fix this using the `.expanduser()` method.

````python
p = Path("~/dotfiles").expanduser()
p
````

You can also chain the `resolve()` method to remove relative bits. 

In addition, this method works only works if the file or directory exists. If we creating it from scratch it would be better to use the `.home() + / "string"`

````python
p = Path.home() / "dotfiles"
p
````

However, this method is limited in terms of searching as extra work is required in accessing information  within the subdirectories of our path. 

A much better method is the `.glob()` method which can be used for more advanced searching. We also note the conditions for using this method:
- `.glob()` — searches the path directories by default.
- `.rglob()` — stands for '_recursive glob_. It searches for subdirectories. 
  
Let's say we want to search our home path (which i will be given an imaginary identifier) for any file or folder with `.vscode` name.

````python
dotfiles = Path.home() 

for p in dotfiles.rglob("*vscode*"):          # The wildcards (*) means 'anything at the beginning or at the end
    print(p)
````

Another caveat worth noting is that the glob module is case-sensitive and so, we may not get the result we desire. We can avoid this situation by adding a keyword argument `case_sensitivity` and setting it to `False`. :

```python
dotfiles = Path.home()        

for p in dotfiles.rglob("*.vscode*", case_sensitive=False):         
    print(p)

```

The wildcards can be used in different ways. You may want to find all the files with a particular extension, say `.json`:

````python
dotfiles = Path.home()        

for p in dotfiles.rglob("*.json", case_sensitive=False):     # No wildcard is needed at the end     
    print(p)
````

In order to open on of the files to work with them, we can pass in a pathlib object into the open function.

For example, let's open the `metadata.json` file in `C:\Users\Dell\.jarvis_profiles\chrome\extensions_crx_cache\metadata.json`. We do not use the glob method here, instead we can open it in this way:

````python
dotfiles = Path.home()  / ".jarvis_profiles\chrome\extensions_crx_cache\metadata.json"     

with dotfiles.open() as f:
    print(f.read())
````

Alternatively, we can also open it as:

````python
dotfiles = Path.home() / ".jarvis_profiles\chrome\extensions_crx_cache\metadata.json"

with open(dotfiles) as f:
    print(f.read())
````

**Note**: _Pathlib objects cna be used with the `.open()` functions directly without the need to convert them to strings first._

With pathlib there are limitations in creating, moving and deleting files and folders.

### Creating Directories
For example, let's create a new directory:

````python
p = Path("TempDir")         # Creates a temporary directory
p.mkdir()
````

````python
# To remove it
p.rmdir()           # This removes only empty directories  
````

We can also create a directory + a sub directory:

````python
p = Path("TempDir/Subdir")
p.mkdir(parents=True)               # This argument creates the non-existent parent directory
````

Now, we cannot delete non-existent directories . However, we can either delete the subdirectory first or use the `shutil` module.

### Creating files
Files are created using the `.touch()` method. If the file already exists, the file will not be overwritten. Instead, its modification time will be updated.

````python
p = Path("TempFile.txt")
p.touch()                   # .touch() only updates the modification time if the file already exists
````

### Renaming files

`.rename()` is used to rename files. However, this might be tricky especially when used across different operating systems. We might need to take precautions by using conditionals to avoid running into errors if the file already exists.

````python
p.rename("tempfile.txt")
````

We can also use the `.replace()` method instead if we want to skip the use of conditionals. 

````python
p = Path("TempFile.txt")   
print(p.replace("temp_file.txt"))
````

### Removing files
`.unlink()` method is used for deleting files. I will be using a set of conditionals just to ensure that this works.

````python
import shutil

p = Path("temp_file.txt")

if p.exists() and p.is_file():
    p.unlink()
    print("Deleted")
elif p.is_dir():
    shutil.rmtree(p)
else:
    print("Doesn't exist")
````
